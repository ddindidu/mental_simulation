```python
from datasets import load_dataset

# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("nvidia/Nemotron-Personas-USA")
```

```python
for row in ds["train"]:
    print(row)
    break
```

```python
# tqdm
from tqdm import tqdm
```

```python
# 모든 컬럼 그대로 유지해서 csv로 변환 (필터링 없음)

rows = []
for row in tqdm(ds["train"]):
    filtered_row = dict(row)
    rows.append(filtered_row)
```

```python
# save to csv
import pandas as pd
df = pd.DataFrame(rows)
df.to_csv("nvidia_persona_filtered.csv", index=False)
```
